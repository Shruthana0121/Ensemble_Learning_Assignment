## Ensemble Learning Assignment

### Theory

**1) Can we use Bagging for regression problems?**

-Yes, Bagging can be used for regression problems. In such cases, it is called a Bagging Regressor, where multiple regression models are trained on different bootstrap samples and their outputs are averaged to produce the final prediction.

**2) What is the difference between multiple model training and single model training?**

-Single model training uses one model to learn patterns from data, while multiple model training (ensemble learning) combines predictions from multiple models to improve accuracy, robustness, and generalization.

**3) Explain the concept of feature randomness in Random Forest?**

-Feature randomness means that at each split in a decision tree, only a random subset of features is considered instead of all features. This reduces correlation between trees and improves model diversity.

**4) What is OOB (Out-of-Bag) Score?**

-OOB score is an internal validation method in Bagging and Random Forest. It evaluates model performance using samples that were not included in the bootstrap training set.

**5) How can you measure the importance of features in a Random Forest model?**

-Feature importance is measured based on how much each feature reduces impurity (like Gini or entropy) across all trees. Higher reduction indicates higher importance.

**6) Explain the working principle of a Bagging Classifier?**

-A Bagging Classifier creates multiple subsets of data using bootstrap sampling, trains a base classifier on each subset, and combines predictions using majority voting.

**7) How do you evaluate a Bagging Classifier’s performance?**

-It can be evaluated using metrics like accuracy, precision, recall, F1-score, ROC-AUC, or cross-validation depending on the problem.

**8) How does a Bagging Regressor work?**

-It trains multiple regression models on different bootstrap samples and combines their outputs by averaging predictions.

**9) What is the main advantage of ensemble techniques?**

-The main advantage is improved accuracy and reduced overfitting by combining multiple models.

**10) What is the main challenge of ensemble methods?**

-The main challenges include increased computational cost, complexity, and reduced interpretability.

**11) Explain the key idea behind ensemble techniques?**

-The key idea is to combine multiple weak learners to create a strong learner that performs better than individual models.

**12) What is a Random Forest Classifier?**

-It is an ensemble learning method that builds multiple decision trees using random subsets of data and features, and combines their outputs using majority voting.

**13) What are the main types of ensemble techniques?**

-The main types are:
- Bagging
- Boosting
- Stacking

**14) What is ensemble learning in machine learning?**

-Ensemble learning is a technique where multiple models are combined to improve prediction performance.

**15) When should we avoid using ensemble methods?**

-Avoid when:
- Dataset is small
- Model interpretability is important
- Computational resources are limited


**16) How does Bagging help in reducing overfitting?**

-Bagging reduces overfitting by averaging multiple models trained on different data samples, thus reducing variance.

**17) Why is Random Forest better than a single Decision Tree?**

-Random Forest reduces overfitting, improves accuracy, and is more robust due to averaging multiple trees.

**18) What is the role of bootstrap sampling in Bagging?**

-Bootstrap sampling creates multiple datasets with replacement, ensuring diversity among models.

**19) What are some real-world applications of ensemble techniques?**

-Applications include:
- Fraud detection
- Medical diagnosis
- Recommendation systems
- Autonomous driving
- Stock prediction


**20) What is the difference between Bagging and Boosting?**

-Bagging trains models independently in parallel, while Boosting trains models sequentially by focusing on previous errors.

**21) Train a Bagging Classifier using Decision Trees on a sample dataset and print model accuracy**

```python
PythonRunfromsklearn.ensembleimportBaggingClassifier
fromsklearn.treeimportDecisionTreeClassifier
fromsklearn.datasetsimportload_iris
fromsklearn.model_selectionimporttrain_test_split
fromsklearn.metricsimportaccuracy_score

data=load_iris()
X_train, X_test, y_train, y_test=train_test_split(data.data, data.target, test_size=0.3)

model=BaggingClassifier(base_estimator=DecisionTreeClassifier(), n_estimators=10)
model.fit(X_train, y_train)

y_pred=model.predict(X_test)
print("Accuracy:", accuracy_score(y_test, y_pred))
```

**22) Train a Bagging Regressor using Decision Trees and evaluate using Mean Squared Error (MSE)**

```python
PythonRunfromsklearn.ensembleimportBaggingRegressor
fromsklearn.treeimportDecisionTreeRegressor
fromsklearn.datasetsimportload_diabetes
fromsklearn.metricsimportmean_squared_error

data=load_diabetes()
X, y=data.data, data.target

model=BaggingRegressor(base_estimator=DecisionTreeRegressor(), n_estimators=10)
model.fit(X, y)

y_pred=model.predict(X)
print("MSE:", mean_squared_error(y, y_pred))
```
